```python
import os
import pandas as pd
import easyocr

# Load metadata CSV
metadata_path = r"C:\Users\Priyanka\OneDrive\Desktop\metadata.csv"

df = pd.read_csv(metadata_path)

print("Total screenshots found:", len(df))

# Create EasyOCR reader
reader = easyocr.Reader(['en'])

# Store OCR results
ocr_texts = []

total_files = len(df)

# Process all screenshots
for i, row in df.iterrows():

    image_path = row["filepath"]

    try:
        result = reader.readtext(image_path, detail=0)
        extracted_text = "\n".join(result)

        ocr_texts.append(extracted_text)

        print(f"{i + 1}/{total_files} completed: {row['filename']}")

    except Exception as e:
        print(f"Error processing {row['filename']}: {e}")
        ocr_texts.append("")

# Add OCR text to the dataframe
df["ocr_text"] = ocr_texts

# Save final CSV
output_path = r"C:\Users\Priyanka\OneDrive\Desktop\Smart_Screenshot_Search_Engine\metadata_with_ocr.csv"

df.to_csv(output_path, index=False, encoding="utf-8")

print("\nOCR COMPLETED SUCCESSFULLY!")
print("Total screenshots processed:", len(df))
print("Saved at:", output_path)

# Display sample
df.head()
```
